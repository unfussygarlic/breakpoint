```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
```

```python
def simulator(theta):
    # one knob: friction. A wiggly curve with a clear global dip (failure).
    # return np.sin(theta) + np.sin((10.0 / 3.0) * theta)
    return np.sin(theta)

DOMAIN = (2.7, 7.5)                       # range of friction values allowed
GRID = np.linspace(*DOMAIN, 400)
n_init = 1000
X = rng.uniform(*DOMAIN, size=n_init).tolist()

sim = [float(simulator(t)) for t in range(len(X))]
```

```python
simulator(10000)
```

```python
plt.scatter(sim, np.arange(len(sim)))
plt.show()
```
